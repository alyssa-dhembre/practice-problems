## 9. Accept 20 values from user and count similar elements/indexes, even/odd, and positive/negative values.

```python
numbers=[]
for i in range(20):
    num=int(input("Enter a number: "))
    numbers.append(num)
print("\nList of numbers:",numbers)
print("\nSimilar elements and their index values:")
checked=[]
for i in range(20):
    if numbers.count(numbers[i])>1 and numbers[i] not in checked:
        print(numbers[i],"-->",end=" ")
        for j in range(20):
            if numbers[i]==numbers[j]:
                print(j,end=" ")
        print()
        checked.append(numbers[i])
even_count=0
odd_count=0
for num in numbers:
    if num%2==0:
        even_count+=1
    else:
        odd_count+=1
print("\nCount of even numbers:",even_count)
print("Count of odd numbers:",odd_count)
positive_count=0
negative_count=0
for num in numbers:
    if num>0:
        positive_count+=1
    elif num<0:
        negative_count+=1
print("\nCount of positive numbers:",positive_count)
print("Count of negative numbers:",negative_count)
```

## 10. Accept 10 values; sort ascending using sorted(), descending using sort(), and display length.

```python
numbers=[]
for i in range(10):
    num=int(input("Enter a number: "))
    numbers.append(num)
ascending=sorted(numbers)
print("Ascending order:",ascending)
numbers.sort(reverse=True)
print("Descending order:",numbers)
print("Length of the list:",len(numbers))
```

## 11. Accept two lists from user and merge them using + in a single list.

## 12. Create an acronym from a phrase.

```python
phrase=input("Enter a phrase: ")
words=phrase.split()
print("Words:",words)
print("Acronym:",end="")
for word in words:
    print(word[0].upper(),end="")
print()
```

## 13. Print the abbreviation of a month, given its number.

## 18. Find common elements in two lists.

```python
list1=[12,24,36,48,60]
list2=[18,24,42,48,72]
common=[]
for num in list1:
    if num in list2:
        common.append(num)
print("List 1:",list1)
print("List 2:",list2)
print("Common elements:",common)
```
